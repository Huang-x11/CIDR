```python
import os
import glob
import json
import numpy as np
import pandas as pd

from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, matthews_corrcoef
```

```python
def get_metrics(json_path):
    with open(json_path) as f:
        data = json.load(f)
        
    pred = np.array(data['pred'])
    pred_oh = pred > 0.5
    label = np.array(data['label'])
    
    acc = accuracy_score(label, pred_oh)
    print('Accuracy: ', round(acc*100, 2))
    
    auc = roc_auc_score(label, pred)
    print('ROC Score: ', round(auc*100, 2))

    f1 = f1_score(label, pred_oh)
    print('F1 Score: ', round(f1*100, 2))

    mcc = matthews_corrcoef(label, pred_oh)
    print('MCC Score: ', round(mcc*100, 2))
```

```python
get_metrics('./logs1/C16_CTRANSPATH/DSMIL/version_0/results.json')
```

```python
df2 = pd.read_csv('../../data/data_c16_ctranspath.csv')

df2['x'] = [item.split('/')[-1].split('.')[0] for item in df2['x']]

train = df2[df2['data_type'] == 'train']['x'].reset_index(drop=True)
train_label = df2[df2['data_type'] == 'train']['y'].reset_index(drop=True)
val = df2[df2['data_type'] == 'valid']['x'].reset_index(drop=True)
val_label = df2[df2['data_type'] == 'valid']['y'].reset_index(drop=True)
test = df2[df2['data_type'] == 'test']['x'].reset_index(drop=True)
test_label = df2[df2['data_type'] == 'test']['y'].reset_index(drop=True)

df = pd.concat([train, train_label, val, val_label, test, test_label], axis=1)
df.columns = ['train', 'train_label', 'val', 'val_label', 'test', 'test_label']

df.to_csv('../backup/TransMIL/dataset_csv/camelyon16/fold1.csv')
```
